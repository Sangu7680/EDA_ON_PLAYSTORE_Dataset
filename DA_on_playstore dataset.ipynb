Data preprocessing


```python
# df.info()

print(df.dtypes)
print(df["Reviews"].dtype)
```

```python
#Convert the datatype of Last Updated to datetime

df['Last Updated'] = pd.to_datetime(df['Last Updated'])

#convert price to numeric
df['Price'] = pd.to_numeric(df['Price'].str.strip('$'),errors="coerce")

#convert reviews to numeric

df['Reviews'] = pd.to_numeric(df['Reviews'],errors="coerce")
```

```python
print(df[df["Last Updated"].isnull()])
print(df[df["Reviews"].isnull()])
```

```python
df.info()
```

```python
#Data Filtering
#Filter by reviews
valid_reviews = df['Reviews']>1000
df = df[valid_reviews]              #Or else you can write df=df[df["Reviews"]>1000] directly without using an extra variable

from datetime import datetime,timedelta

#Filter by last updates
seven_years_ago = datetime.now() - timedelta(days= 365*7)
valid_updates = df['Last Updated'] >= seven_years_ago

df = df[valid_updates]

#Filter by installs 
valid_installs = df['Installs'] > 50000
df = df[valid_installs]

#Filter by rating
valid_rating = df['Rating']>2.0
df = df[valid_rating]

df


```

```python
# df.drop(df.columns[[13]],axis=1,inplace=True)
df
```

```python
df.info()
```

```python
df.describe()
```

```python
#Scaling the reviews and Installs

df['Installs'] = df['Installs']/10000

df['Reviews'] = df['Reviews']/1000
```

```python
df.to_csv("googleplaystore_cleaned.csv",index=False)
```

Visualisation

```python
#Lets find out number of games category wise

# plt.figure(figsize=(50,5))
# sns.histplot(df['Category'])
# plt.xlabel('Category')

# or For better  understandment you can use  --

plt.figure(figsize=(25, 10))  
sns.set_style("whitegrid")     #or you can write sns.set(style="whitegrid")
plt.rc('axes', labelsize=22)  
plt.rc('xtick', labelsize=18)
plt.rc('ytick', labelsize=18)

custom_palette = sns.color_palette("viridis", n_colors=len(df['Category'].unique()))  # Use `n_colors` for uniqueness
ax = sns.countplot(data=df, x='Category', palette=custom_palette)

plt.xlabel('Game Category', fontsize=22, weight='bold')
plt.ylabel('Number of Games', fontsize=22, weight='bold')
plt.title('Game Counts by Category', fontsize=26, weight="bold", pad=20)

plt.xticks(rotation=45, ha='right')

for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=16, padding=5, weight='bold')

plt.show()
```

```python
#content rating visualisation

plt.figure(figsize=(14, 8))  
sns.set_style("whitegrid")   
plt.rc('axes', labelsize=12)
plt.rc('xtick',labelsize=12)
plt.rc('ytick',labelsize=12)

palette = sns.color_palette("pastel") 
ax = sns.countplot(data=df, x='Content Rating', palette=palette)

plt.xlabel('Content Rating', fontsize=14, weight='bold')
plt.ylabel('Frequency', fontsize=14, weight='bold')
plt.title('Distribution of Content Ratings', fontsize=18, weight='bold', pad=20)

plt.xticks(rotation=45, ha='left')

for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=10, padding=3, weight='bold')

plt.show()
```

```python
#Genres visualisation

genre_counts = df['Genres'].value_counts()

df['Genres'] = df['Genres'].apply(lambda x: x if genre_counts[x] >= 30 else 'Others')

plt.figure(figsize=(16, 8))
sns.set_style("whitegrid")
plt.rc('axes', labelsize=12)
plt.rc('xtick', labelsize=12)
plt.rc('ytick', labelsize=12)

sns.histplot(data=df, x='Genres', color='skyblue', edgecolor='black', linewidth=1.2)

plt.xlabel('Genres', fontsize=14, weight='bold')
plt.ylabel('Frequency', fontsize=14, weight='bold')
plt.title('Distribution of Genres (Grouped)', fontsize=18, weight='bold', pad=20)

plt.xticks(rotation=45, ha='right')

plt.show()

```

```python
#Distribution of App rating

sns.set_style("whitegrid") 
plt.figure(figsize=(10, 6))
sns.histplot(df['Rating'].dropna(), bins=20, kde=True, color='dodgerblue', edgecolor='black')
plt.title('Distribution of App Ratings', fontsize=16, fontweight='bold')
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.show()
```

```python
#Average rating per category

category_ratings = df.groupby('Category')['Rating'].mean()

plt.figure(figsize=(14, 8))
sns.barplot(x=category_ratings.index, y=category_ratings.values, data=df,palette='coolwarm', edgecolor='black')

plt.title('Average Rating per Category', fontsize=16, fontweight='bold')
plt.xlabel('Category', fontsize=12)
plt.ylabel('Average Rating', fontsize=12)
plt.xticks(rotation=90, fontsize=10)
plt.yticks(fontsize=10)

for i, v in enumerate(category_ratings.values):
    plt.text(i, v + 0.02, f'{v:.2f}', ha='center', fontsize=10)

plt.show()
```

```python
#Ratings according to reviews

plt.figure(figsize=(10, 6))
plt.scatter(df['Rating'], df['Reviews'].astype(float), c=df['Rating'], cmap='viridis', alpha=0.6, edgecolor='w', s=50)

plt.title('Rating vs. Number of Reviews', fontsize=16, fontweight='bold')
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Number of Reviews', fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)

plt.colorbar(label='Rating')
plt.show()
```
